```python
import pandas as pd
import numpy as np
import duckdb
import json
from pathlib import Path

# Locate Project Paths
PROJECT_ROOT = Path("..").resolve()
RAW_CSV = PROJECT_ROOT / "data" / "raw" / "messy_students.csv"
CLEAN_CSV = PROJECT_ROOT / "data" / "processed" / "cleaned_students.csv"
QUALITY_JSON = PROJECT_ROOT / "data" / "processed" / "quality_report.json"
DUCKDB_PATH = PROJECT_ROOT / "data" / "processed" / "analytics.duckdb"
```

## 1. Inspecting Raw Messy Dataset

```python
df_raw = pd.read_csv(RAW_CSV)
print(f"Raw records: {len(df_raw)}")
df_raw.head(10)
```

## 2. Reviewing Data Rescue Audit Metrics

```python
with open(QUALITY_JSON, "r", encoding="utf-8") as f:
    report = json.load(f)

print("=== DATA RESCUE QUALITY SCORE ===")
print(f"Data Quality Score: {report.get('data_quality_score')}/100")
print(f"Duplicates Dropped: {report.get('duplicates_removed')}")
print(f"Missing Values Imputed: {report.get('missing_values_imputed')}")
print(f"Invalid CGPA Repaired: {report.get('invalid_cgpa_fixed')}")
print(f"Attendance Normalized: {report.get('attendance_normalized')}")
print(f"Suspicious Outliers Detected: {report.get('suspicious_records_detected')}")
```

## 3. Querying Embedded DuckDB Analytics

```python
con = duckdb.connect(str(DUCKDB_PATH), read_only=True)
dept_summary = con.execute("SELECT * FROM department_summary ORDER BY avg_cgpa DESC;").df()
dept_summary
```
